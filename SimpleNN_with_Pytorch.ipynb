```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
import numpy as np
```

```python
data=load_breast_cancer()
```

```python
data
```

```python
df=pd.DataFrame(data=data.data,columns=data.feature_names)
```

```python
df['Target']=pd.DataFrame(data=data.target)
```

```python
df.sample(7)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop('Target',axis=1),df['Target'],test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
x_train
```

```python
#Numpy arrays to tensor
```

```python
import torch
```

```python
x_train_tensor=torch.from_numpy(x_train)
x_test_tensor=torch.from_numpy(x_test)
y_train_tensor=torch.from_numpy(y_train.values)
y_test_tensor=torch.from_numpy(y_test.values)
```

```python
x_train_tensor
```

```python
x_train_tensor.shape
```

# Model

```python
class SimpleNN():
  def __init__(self,X):
    torch.manual_seed(42)
    self.weights=torch.rand(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias=torch.zeros(1,1,dtype=torch.float64,requires_grad=True)

  def forward(self,X):
    z= torch.matmul(X,self.weights) + self.bias
    y_pred=torch.sigmoid(z)
    return y_pred

  def loss_cal(self,y_pred,y):
    epsilon=1e-7
    y_pred=torch.clamp(y_pred,epsilon,1-epsilon)

    loss=-(y_train_tensor*torch.log(y_pred) + (1-y_train_tensor)*torch.log(1-y_pred))
    return loss.mean()

```

```python
learning_rate=0.1
epochs=25
```

# Pipeline

```python
model=SimpleNN(x_train_tensor)
model.weights
```

```python
#loop
for epoch in range(epochs):
  #Forward
  y_pred=model.forward(x_train_tensor)

  #Loss
  loss=model.loss_cal(y_pred,y_train_tensor)

  #Backprop
  loss.backward()

  #Update weights
  with torch.no_grad():
    model.weights-=learning_rate*model.weights.grad
    model.bias-=learning_rate*model.bias.grad

  #since gradients accumulate zero them
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  print(f"Epoch:{epoch},loss={loss.item()}")

```

Model evaluation


```python
with torch.no_grad():
  y_pred=model.forward(x_test_tensor)
# original target is either 0 or 1,so we need to conver y_pred to 0 or 1 with threshold 0.5
  y_pred=(y_pred>0.5).float()

  #Accuracy
  accuracy=(y_pred==y_test_tensor).float().mean()
  print(f"Accuracy:{accuracy}")
```
